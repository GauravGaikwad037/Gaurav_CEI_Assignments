# Creating Helper Function

```python
def run_query(query):
    return pd.read_sql_query(query, conn)
```

# Applying Subqueries

## Q1 Orders With Sales Greater Than Average Sales

```python
run_query("""
SELECT *

FROM superstore_raw

WHERE Sales >
(
SELECT AVG(Sales)

FROM superstore_raw
);
""")
```

## Q2 Highest Order Of Each Customer

```python
run_query("""
SELECT *

FROM superstore_raw s

WHERE Sales =
(
SELECT MAX(Sales)

FROM superstore_raw

WHERE "Customer ID" = s."Customer ID"
);
""")
```

# Common Table Expressions (CTEs)

## Q3 Total Sales Per Customer Using CTE

```python
run_query("""
WITH CustomerSales AS
(
SELECT

"Customer ID",

"Customer Name",

SUM(Sales) AS Total_Sales

FROM superstore_raw

GROUP BY

"Customer ID",

"Customer Name"
)

SELECT *

FROM CustomerSales

ORDER BY Total_Sales DESC;
""")
```

## Q4 Customers Above Average Total Sales

```python
run_query("""
WITH CustomerSales AS
(
SELECT

"Customer ID",

"Customer Name",

SUM(Sales) AS Total_Sales

FROM superstore_raw

GROUP BY

"Customer ID",

"Customer Name"
)

SELECT *

FROM CustomerSales

WHERE Total_Sales >
(
SELECT AVG(Total_Sales)

FROM CustomerSales
);
""")
```

# Window Functions

## Q5 Assign Row Number Based On Sales

```python
run_query("""
SELECT

"Customer Name",

Sales,

ROW_NUMBER()

OVER
(
ORDER BY Sales DESC
)

AS Row_Number

FROM superstore_raw;
""")
```

## Q6 Ranking Customers Based On Sales

```python
run_query("""
SELECT

"Customer Name",

Sales,

RANK()

OVER
(
ORDER BY Sales DESC
)

AS Sales_Rank

FROM superstore_raw;
""")
```

# Business Queries

## Q7 Top 10 Customers Based on Total Sales

```python
run_query("""
WITH CustomerSales AS
(
SELECT

"Customer ID",

"Customer Name",

SUM(Sales) AS Total_Sales

FROM superstore_raw

GROUP BY

"Customer ID",

"Customer Name"
)

SELECT *

FROM CustomerSales

ORDER BY Total_Sales DESC

LIMIT 10;
""")
```

## Q10 Ranking Customers Based on Total Sales

```python
run_query("""
WITH CustomerSales AS
(
SELECT

"Customer ID",

"Customer Name",

SUM(Sales) AS Total_Sales

FROM superstore_raw

GROUP BY

"Customer ID",

"Customer Name"
)

SELECT

*,

RANK()

OVER
(
ORDER BY Total_Sales DESC
)

AS Customer_Rank

FROM CustomerSales;
""")
```

## Q11 Customer Sales Ranking Using JOIN, CTE and Window Function

```python
run_query("""
WITH CustomerSales AS
(
SELECT

"Customer ID",

SUM(Sales) AS Total_Sales

FROM superstore_raw

GROUP BY "Customer ID"
)

SELECT

c."Customer ID",

c."Customer Name",

cs.Total_Sales,

RANK()

OVER
(
ORDER BY cs.Total_Sales DESC
)

AS Customer_Rank

FROM customers c

JOIN CustomerSales cs

ON c."Customer ID" = cs."Customer ID"

ORDER BY Customer_Rank;
""")
```
